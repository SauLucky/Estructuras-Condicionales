###2 - Escribe un programa que solicite el porcentaje de crecimiento de producción de una empresa e informe si hubo un crecimiento (porcentaje positivo) o una disminución (porcentaje negativo).


```python
print("Ingresa el porcentaje de produccion de los siguientes años")
produccion_2023 = float(input("Ingresa el porcentaje de produccion de 2023: "))
produccion_2024 = float(input("Ingresa el porcentaje de produccion de 2024: "))

# Calcula la diferencia entre los porcentajes
diferencia = produccion_2024 - produccion_2023

# Evalúa si la diferencia es positiva o negativa
if diferencia > 0:
  print("Hubo un crecimiento en la produccion del", diferencia, "%")
elif diferencia < 0:
  print("Hubo un disminución en la produccion del", (diferencia), "%")
else:
  print("No hubo cambio en la produccion.")
```

###4 - Escribe un programa que lea valores promedio de precios de un modelo de automóvil durante 3 años consecutivos y muestre el valor más alto y más bajo entre esos tres años.


```python
print("Ingresa el valor del Versa Base en los siguientes años:")

precio_2023 = float(input("Ingresa el valor del Versa Base en 2023: "))
precio_2024 = float(input("Ingresa el valor del Versa Base en 2024: "))
precio_2025 = float(input("Ingresa el valor del Versa Base en 2025: "))

# Encontrar el valor más alto y más bajo
valor_mas_alto = max(precio_2023, precio_2024, precio_2025)
valor_mas_bajo = min(precio_2023, precio_2024, precio_2025)

# Encontrar el año del valor más alto y más bajo
if valor_mas_alto == precio_2023:
    año_mas_alto = 2023
elif valor_mas_alto == precio_2024:
    año_mas_alto = 2024
else:
    año_mas_alto = 2025

if valor_mas_bajo == precio_2023:
    año_mas_bajo = 2023
elif valor_mas_bajo == precio_2024:
    año_mas_bajo = 2024
else:
    año_mas_bajo = 2025

print(f"El valor más alto fue de ${valor_mas_alto} en el año {año_mas_alto}")
print(f"El valor más bajo fue de ${valor_mas_bajo} en el año {año_mas_bajo}")
```

##11 - Escribe un programa que pida a la persona usuaria tres números que representan los lados de un triángulo. El programa debe informar si los valores pueden utilizarse para formar un triángulo y, en caso afirmativo, si es equilátero, isósceles o escaleno. Ten en cuenta algunas sugerencias:

*   Tres lados forman un triángulo cuando la suma de cualesquiera dos lados es mayor que el tercero;
*   Triángulo Equilátero: tres lados iguales;
*   Triángulo Isósceles: dos lados iguales;
*   Triángulo Escaleno: tres lados diferentes.

```python
print('Ingresa tres lados de un triangulo')
lado1 = float(input('Ingresa el primer lado: '))
lado2 = float(input('Ingresa el segundo lado: '))
lado3 = float(input('Ingresa el tercer lado: '))
if lado1 + lado2 > lado3 and lado1 + lado3 > lado2 and lado2 + lado3 > lado1:
  print('Los lados ingresados pueden formar un triangulo')
  if lado1 == lado2 == lado3:
    print('El triangulo es equilatero')
  elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print('El triangulo es isoceles')
  else:
    print('El triangulo es escaleno')
else:
  print('Los lados ingresados no pueden formar un triangulo')

```

##12 - Un establecimiento está vendiendo combustibles con descuentos variables.

###Para el etanol, si la cantidad comprada es de hasta 15 litros, el descuento será del 2% por litro. En caso contrario, será del 4% por litro. Para el diésel, si la cantidad comprada es de hasta 15 litros, el descuento será del 3% por litro. En caso contrario, será del 5% por litro. El precio por litro de diésel es de &dollar;2.00 y el precio por litro de etanol es de &dollar;1.70. Escribe un programa que lea la cantidad de litros vendidos y el tipo de combustible (E para etanol y D para diésel) y calcule el valor a pagar por el cliente. Ten en cuenta algunas sugerencias:


###El valor del descuento será el producto del precio por litro, la cantidad de litros y el valor del descuento.
###El valor a pagar por un cliente será el resultado de la multiplicación del precio por litro por la cantidad de litros menos el valor del descuento resultante del cálculo.

```python
print('Ingresa la cantidad de litros vendidos y el tipo de combustible (E para etanol y D para diésel)')
input_cantidad = float(input('Ingresa la cantidad de litros vendidos: '))
input_tipo = input('Ingresa el tipo de combustible (E para etanol y D para diésel): ').upper()

if input_tipo == 'E':
    precio_litro = 1.70
    if input_cantidad <= 15:
        descuento = 0.02
        valor_descuento = precio_litro * input_cantidad * descuento
    else:
        descuento = 0.04
        valor_descuento = precio_litro * input_cantidad * descuento
elif input_tipo == 'D':
    precio_litro = 2.0
    if input_cantidad <= 15:
        descuento = 0.03
        valor_descuento = precio_litro * input_cantidad * descuento
    else:
        descuento = 0.05
        valor_descuento = precio_litro * input_cantidad * descuento
else:
    print('Tipo de combustible no válido')
valor_a_pagar = precio_litro * input_cantidad - valor_descuento
print(f'El valor a pagar por el cliente es:${valor_a_pagar}')
```

##13 - En una empresa de venta de bienes raíces, debes crear un código que analice los datos de ventas anuales para ayudar a la dirección en la toma de decisiones.
El código debe recopilar los datos de cantidad de ventas durante los años 2022 y 2023 y calcular la variación porcentual. A partir del valor de la variación, se deben proporcionar las siguientes sugerencias:


*   Para una variación superior al 20%: bonificación para el equipo de ventas.
*   Para una variación entre el 2% y el 20%: pequeña bonificación para el equipo de ventas.
*   Para una variación entre el 2% y el -10%: planificación de políticas de incentivo a las ventas.
*   Para bonificaciones inferiores al -10%: recorte de gastos.




```python
print("Ingrese el valor de las ventas anuales de los años 2022 y 2023")
ventas_2022= float(input("Ventas 2022:$ "))
ventas_2023= float(input("Ventas 20223: $"))
variacion = ((ventas_2023 - ventas_2022) / ventas_2022) * 100
print("La variacion porcentual es del", variacion,"%")
if variacion > 20:
  print("Se sugiere: Bonificacion para el equipo de ventas")
elif variacion >= 2 and variacion <= 20:
  print("Se sugiere: Pequeña bonificacion para el equipo de ventas")
elif variacion >= -10 and variacion < 2:
  print("Se sugiere: Planificacion de politicas de incentivo a las ventas")
else:
  print("Se sugiere: Recorte de gastos")
```
